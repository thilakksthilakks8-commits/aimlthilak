```python
#global scope
x = 300

def myfunc():
  print(x)

myfunc()

print(x)
```

```python
#local scope
def myfunc():
  x = 300
  print(x)

myfunc()
```

```python
#both local global variable
x = 300

def myfunc():
  global x
  x = 200

myfunc()

print(x)
```

```python
#lambda
x = lambda a : a + 10
print(x(5))
```

```python
#2arguments
x = lambda a, b : a * b
print(x(5, 6))
```

```python
#Summarize argument a, b, and c and return the result:

x = lambda a, b, c : a + b + c
print(x(5, 6, 2))
```

```python
#Summarize argument a, b, and c and return the result:

x = lambda a, b, c : a * b + c
print(x(5, 6, 2))
```

```python
#usage of lambda functions
def myfunc(n):
  return lambda a : a * n

mydoubler = myfunc(2)

print(mydoubler(11))
```

```python
def myfunc(n, p):
  return lambda a,b:a* n +  p*b

mydoubler = myfunc(2,3)

print(mydoubler(11,12))
```

```python
#Sort strings by length:

words = ["apple", "pie", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
```
